# Proyecto Integrador de Aprendizaje Automático

Para las empresas financieras se sabe que poder predecir el riesgo de incumplimiento en préstamos es importante para minimizar pérdidas y ayudas en decisiones de la vida crediticia del sujeto. Por ello, nos vamos a enfoca en construir un modelo de clasificación en el que podamos interpretar los préstamos que serán pagados en su totalidad o parcial, cuánto quedarán debiendo y/o aquellos que estarían en default.

```python
import pandas as pd

archivo_entrada = "data_proyect.csv"
archivo_salida = "data_proyect_limpio.csv"

# Definir tamaño de los chunks (ajústalo según memoria disponible)
chunksize = 10_000  

# Abrimos archivo de salida en modo escritura y guardamos encabezado
with open(archivo_salida, "w", encoding="utf-8") as f_out:
    for i, chunk in enumerate(pd.read_csv(archivo_entrada, sep=",", chunksize=chunksize, low_memory=False)):
        # Procesamos cada bloque (ejemplo: eliminar filas vacías)
        chunk = chunk.dropna(how="all")

        if i == 0:
            # Primera vez: escribimos con encabezados
            chunk.to_csv(f_out, index=False)
        else:
            # Resto: escribimos sin encabezados
            chunk.to_csv(f_out, index=False, header=False)
```
